## Pesquisa Diaria ##

```python
import pyautogui
import time
import random

search = [
    "Star Wars","Harry Potter","Avatar","Hobbit","Senhor dos Aneis",
    "Homem-Aranha","Batman","Superman","Vingadores","Homem de Ferro",
    "Capitao America","Thor","Hulk","Pantera Negra","Doutor Estranho",
    "Flash","Liga da Justica","Deadpool","X-Men","Wolverine",
    "Naruto","Dragon Ball","One Piece","Attack on Titan","Demon Slayer",
    "Jujutsu Kaisen","Death Note","Tokyo Ghoul","Fullmetal Alchemist","Bleach",
    "Minecraft","Fortnite","GTA 5","Red Dead Redemption 2","The Last of Us",
    "God of War","Call of Duty","FIFA","Free Fire","League of Legends",
    "Valorant","CS GO","Roblox","Among Us","Terraria","Cyberpunk 2077",
    "Elden Ring","Dark Souls","Skyrim","The Witcher","Stranger Things",
    "Breaking Bad","Game of Thrones","The Walking Dead","Peaky Blinders",
    "The Boys","Wednesday serie","Loki serie","Wandavision","Black Mirror",
    "Rick and Morty","Simpsons","Family Guy","South Park","filmes 2025",
    "melhores series","noticias tecnologia","inteligencia artificial",
    "como programar","python tutorial","jogos novos","filmes lancamento",
    "curiosidades espaco","buracos negros","planetas do sistema solar",
    "viagem no tempo","teoria da relatividade","historia do Brasil",
    "segunda guerra mundial","imperio romano","mitologia grega","dinossauros",
    "animais perigosos","como estudar melhor","dicas produtividade",
    "como ganhar dinheiro online","negocios digitais","marketing digital",
    "como fazer exercicios","treino em casa","alimentacao saudavel",
    "receitas faceis","como dormir melhor"
]

extra = ["curiosidades", "historia", "explicacao", "teoria", "review"]

# gera combinações únicas e embaralha
combinations = [f"{t} {e}" for t in search for e in extra]
random.shuffle(combinations)

def delay_humano():
    t = random.uniform(1.2, 2.5)

    # às vezes demora mais (tipo pessoa distraída)
    if random.random() < 0.25:
        t += random.uniform(1, 3)

    return t

time.sleep(1)

# abrir navegador / barra
time.sleep(1.5)
pyautogui.click(x=264, y=22)
pyautogui.click(x=600, y=60)
pyautogui.write("https://www.bing.com/?FORM=Z9FD1")
time.sleep(1)
pyautogui.press("enter")

# primeira busca
time.sleep(2)
pyautogui.click(x=880, y=325)
pyautogui.write(combinations[0])
pyautogui.press("enter")

# restante
for atual in combinations[1:91]:
    time.sleep(delay_humano())

    pyautogui.click(x=355, y=165)
    pyautogui.hotkey("ctrl", "a")
    pyautogui.press("backspace")

    pyautogui.write(atual, interval=.05)
    pyautogui.press("enter")
```
